```python
import matplotlib.pyplot as plt
import numpy as np
with open ('iris_data.csv', 'r') as f:
    lines = f.readlines()
a1 = 0
a2 = 0
a3 = 0
b1 = 0
b2 = 0
b3 = 0
for line in lines:
    row = line.strip().split(',')
    if row[-1] == 'Iris-setosa':
        a1 += 1
    if row[-1] == 'Iris-versicolor':
        a2 += 1
    if row[-1] == 'Iris-virginica':
        a3 += 1
    if float(row[-3]) < 1.2:
        b1 += 1       
    if 1.2 < float(row[-3]) < 1.5 :
        b2 += 1
    if float(row[-3]) > 1.5:
        b3 += 1
    
plt.pie([a1/(a1+a2+a3), a2/(a1+a2+a3), a3/(a1+a2+a3)], labels = ['Iris-setosa','Iris-versicolor','Iris-virginica'])

plt.title('Доли разных видов')
plt.show()

plt.pie([b1, b2, b3], labels = ['Меньше 1.2','Больше 1.2 и меньше 1.5','Больше 1.5'])

plt.title('Доли разных размеров')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
with open ('iris_data.csv', 'r') as f:
    lines = f.readlines()
sl = []
sw = []
pl = []
pw = []
for line in lines:
    row = line.strip().split(',') 
    sl.append(float(row[1]))
    sw.append(float(row[2]))
    pl.append(float(row[3]))
    pw.append(float(row[4]))
plt.scatter(sw, sl, marker='.')
plt.xlabel('Sepal Width')
plt.ylabel('Sepal Lenght')
plt.show()

plt.scatter(pl, sl, marker='.')
plt.xlabel('Petal Lenght')
plt.ylabel('Sepal Lenght')
plt.show()

plt.scatter(pw, sl, marker='.')
plt.xlabel('Petal Width')
plt.ylabel('Sepal Lenght')
plt.show()

plt.scatter(pl, sw, marker='.')
plt.xlabel('Petal Lenght')
plt.ylabel('Sepal Width')
plt.show()

plt.scatter(pw, sw, marker='.')
plt.xlabel('Petal Width')
plt.ylabel('Sepal Width')
plt.show()

plt.scatter(pw, pl, marker='.')
plt.xlabel('Petal Width')
plt.ylabel('Petal Lenght')
plt.show()
    
```
